## MC integration

Calculate the integral $\int_0^1 x^n \sin(\alpha\pi x){\rm d}x$ for $n=20$, $\alpha=20.5$ using the Monte Carlo method with 1000 uniformly distributed samples.


```python
def f(x, n=20, alpha=20.5): return x**n*np.sin(np.pi*alpha*x)
```

```python
x = np.linspace(0, 1, 200)
plt.plot(x, f(x))
```

```python
def integrate_MC(f, a, b, nsampl):
    xsamples = uniform(a, b, nsampl)
    fsamples = f(xsamples)
    integral = np.mean(fsamples)*(b-a)
    return integral
```

```python
integrate_MC(f, 0, 1, 1000)
```

Estimate the error of the calculation. Repeat for 1, 10, 100,.... 100000 samples

```python
def test_integrate_MC(f, a, b, nsampl, ntest, integrator=integrate_MC):
    integrals = [integrator(f, a, b, nsampl) for _ in range(ntest)]
    return np.mean(integrals), np.std(integrals)
```

```python
nsampls = 10**np.arange(6)
ntest = 20
```

```python
integrals = np.array([test_integrate_MC(f, 0, 1, nsampl, ntest) for nsampl in nsampls])
```

```python
plt.plot(nsampls, integrals[:,0], "-o")
plt.gca().set_xscale("log")
plt.grid()
```

```python
plt.errorbar(nsampls, integrals[:,0], fmt="-o", yerr=integrals[:,1]/np.sqrt(ntest))
plt.gca().set_xscale("log")
plt.grid()
```

```python
plt.loglog(nsampls, integrals[:,1], "-o")
plt.gca().set_xscale("log")
plt.grid()
```

### Random Number Generation - inverse CDF

Generate random numbers with distribution function $f = (n+1)x^n$ on the interval 0, 1 and verify the functionality using a histogram

note: the cumulative distribution function method is based on the transformation of random variables. Let $x$ be from the distribution $f(x)dx$:
$$x\sim f(x)dx$$
what will be the distribution of the variable $y = g(x)$? We perform the substitution as when changing variables in an integral, i.e. we use:
$$x = g^{-1}(y),\quad dx = \frac{dg^{-1}}{dy}dy$$
and we get
$$f(x)dx = f(g^{-1}(y))\frac{dg^{-1}}{dy}dy = f'(y)dy$$
where the transformed probability density is $f'(y) = f(g^{-1}(y))\frac{dg^{-1}}{dy}$.

If we want to generate $y\sim h(y)dy$ by transforming the quantity $x\sim U(0,1)dx$,
$$f(x) = 1 {\ \rm for\ } x\in \left<0,1\right>, {\ \rm otherwise\ } 0$$
and we use the inverse distribution function as the transformation function, i.e. $g = H^{-1}; H(y) = \int_{-\infty}^y h(y')dy'$, so we get
$$ g^{-1} = H;\quad \frac{dg^{-1}}{dy} = h(y)$$
and
$$f(H(y)) = 1 \forall y$$
therefore
$$f'(y) = h(y)$$


```python
def gen_dist(x, n=20):
    return x**n*(n+1)

def gen_inverse(nsampl, n=20):
    # DF = x**n*(n+1)
    # CDF = x**(n+1)
    # inv CDF = x**(1/(n+1))
    xsampl = uniform(0, 1, nsampl)
    return xsampl**(1/(n+1))
```

```python
samples = gen_inverse(10000)
```

```python
plt.hist(samples, density=True, bins=100);
x = np.linspace(0,1)
plt.plot(x, x**20*21)
```

### Importance sampling

```python
def integrate_MC_importance(f, a, b, nsampl):
    xsamples = gen_inverse(nsampl)
    fsamples = f(xsamples)/gen_dist(xsamples)
    integral = np.sum(fsamples)/nsampl*(b-a)
    return integral
```

```python
integrate_MC_importance(f, 0, 1, 1000)
```

```python
integrals_importance = np.array([test_integrate_MC(f, 0, 1, nsampl, ntest, integrator=integrate_MC_importance) for nsampl in nsampls])
```

```python
plt.loglog(nsampls, integrals[:,1], "-o")
plt.loglog(nsampls, integrals_importance[:,1], "-o")
plt.gca().set_xscale("log")
plt.grid()
```

### Scattering - random direction

```python
def random_direction(nsampl, l=1):
    """generate nsampl 3D vectors with random directions and length l"""
    v = np.zeros((nsampl, 3))
    cos_theta = 1 - 2*np.random.uniform(size = nsampl)
    sin_theta = np.sqrt(1-cos_theta**2)
    phi = 2*np.pi*np.random.uniform(size = nsampl)

    v[:,0] = l*cos_theta
    v[:,1] = l*sin_theta*np.sin(phi)
    v[:,2] = l*sin_theta*np.cos(phi)

    return v
```

```python
%%time
vs = random_direction(1000000)
```

```python
plt.hist2d(vs[:,0], vs[:,1], bins=100)
plt.gca().set_aspect("equal")
```

```python
from numba import njit

@njit
def random_direction_rejection_njit(nsampl, l=1):
    """generate nsampl 3D vectors with random directions and length l"""
    v = np.zeros((nsampl, 3))
    for i in range(nsampl):
        while True:
            r = np.random.uniform(0, 1, size = 3)*2 - 1
            rnorm = np.linalg.norm(r)
            if rnorm <= 1:
                break
        v[i, :] = r/rnorm
    return v
```

```python
%%time
vs = random_direction_rejection_njit(1000000)
```

```python
plt.hist2d(vs[:,0], vs[:,1], bins=100)
plt.gca().set_aspect("equal")
```
